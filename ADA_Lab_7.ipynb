# Question 1

**1. You roll a fair six-sided die. What is the probability of rolling a 3 on the first roll and a 5 on the second roll? (Assume each roll is independent).**

Knowns:
1. $P\text{(3 on 1st roll) = 1/6.}$
  - 1/6 chance of rolling a 3 on a fair six-sided die.
2. $P\text{(5 on 2nd roll) = 1/6.}$
  - 1/6 chance of rolling a 5 on a fair six-sided die.

$P(\text{3 on 1st roll} \land \text{5 on 2nd roll})$  
> $= P(\text{3 on 1st roll}) \times P(\text{5 on 2nd roll})$  
> $= 1/6 \times 1/6$  
> $= 1/36$

# Question 2

**You have a bag with six marbles: three red marbles and three blue marbles. If you draw a marble, record its color, and then do not put it back into the bag before drawing another marble, what is the probability of the following?**

Knowns:
1. 6 total marbles.
   - 3 red, 3 blue.
2. Marbles are not replaced after picking.

**1. What is the probability of drawing a red marble on the first draw?**

> Assuming the marbles *are* uniform, drawing a red on the first draw would be a 3/6 chance, or **1/2**.

**2. What is the probability of drawing a blue marble on the second draw after drawing a red marble on the first draw?**

> Without replacing the red marble previously picked, the chance of drawing a blue marble is **3/5** since three of the five remaining marbles in the bag are blue.

**3. Calculate the probability of both events occuring (drawing a red marble on the first draw, without replacement, and a blue marble on the second draw).**

> The probability of both events occuring would be $P(\text{Blue | Red}) = P(\text{Red on 1st}) \times 3/5 = 3/6 \times 3/5$, i.e., **0.3**.

# Question 3

**You flip a fair coin. What is the probability of getting heads or tails?**

Knowns:
1. $P\text{(Heads) = 1/2.}$
  - 1/2 chance of getting heads on a fair coin.
2. $P\text{(Tails) = 1/2.}$
  - 1/2 chance of getting tails on a fair coin.

$P(\text{Heads} \lor \text{Tails})$
> $= P(\text{Heads}) + P(\text{Tails})$  
> $= 1/2 + 1/2$  
> $= 1$  

You will always roll either heads or tails on a fair coin.

# Question 4

**In a deck of 52 cards, what is the probability of drawing either a spade or a face card, i.e., a king, queen, or jack?**

Knowns:
1. $P\text{(Spade) = 13/52.}$
  - 13 spades in a standard 52-card deck: {♠A, ♠2, ♠3, ..., ♠K}.
2. $P\text{(Face Card) = 12/52.}$
  - 12 face cards in a 52-card deck: {♠K, ♠Q, ♠J, ♥K, ♥Q, ♥J, ♣K, ♣Q, ♣J, ♦K, ♦Q, ♦J}.
3. $P\text{(Spade Face Card) = 3/52.}$
  - 3 face cards that are also spades: {♠K, ♠Q, ♠J}. Since the King, Queen, and Jack of spades are each accounted for in either of the previous sets, they are **not** included in the probability an additional time.

$P(\text{Spade} \lor \text{Face Card})$
> $= P(\text{Spade}) + P(\text{Face Card}) - P(\text{Spade Face Card})$  
> $= 13/52 + 12/52 - 3/52$  
> $= 22/52$
