1. A school asserted that its students are more intelligent than the average student population. To test this, the IQ scores of 50 students were analyzed, resulting in an average score of 110. Given that the population mean IQ is 100 with a standard deviation of 15, determine whether the principal's claim is valid at a 5% significance level.


```python
import numpy as np
import scipy.stats as stats

# 1. Formulate hypothesis
# null hypothesis: the average IQ of the school's students are the same as the national average of 100.
# alt. hypothesis: the average IQ of the school's students is greater than the national average of 100.
# RIGHT-tailed test

# 2. Identify sample values
# sample mean: 110 points
sample_mean = 110

# national average (pop. mean): 100 points
population_mean = 100

# pop. standard deviation: 15 points
population_std = 15

# sample size: 50 students
sample_size = 50


# 3. Calculate the z-score
z_score = (sample_mean - population_mean)/(population_std/np.sqrt(sample_size))
print(f"z-score: {round(z_score, 10)}")

# 4. Compare z-score with significance level or p-value
# significance level 5%
alpha = 0.05

# p-value
p_value = 1-stats.norm.cdf(z_score)
print(f"p value: {round(p_value, 10)}")

# 5. Interpret the results
if (p_value < alpha):
  print("\nP-value < alpha")
  print("REJECT the null hypothesis:")
  print("There IS significant evidence to support the claim that the average IQ of the school's students is greater than the national average of 100.")
else:
  print("\nP-value >= alpha")
  print("FAIL to reject the null hypothesis:")
  print("There is NOT significant evidence to support the claim that the average IQ of the school's students is greater than the national average of 100.")
```

2. A company claims that the average response time for their customer support hotline is less than 5 minutes. You conducted a sample of 36 calls and found that the average response time was 4.7 minutes with a standard deviation of 1.2 minutes. Conduct a hypothesis test to determine whether there is sufficient evidence to support the company's claim.

```python
# 1. null hypothesis: The average response time for the company's customer support hotline is 5 minutes or greater.
# alt. hypothesis: The average  response time for the company's customer support hotline is less than 5 minutes.
# LEFT-tailed test

# 5-minute response time
null_mean = 5

# sample size: 36 calls
sample_size = 36
# sample mean: 4.7 minutes
sample_mean = 4.7
# sample standard deviation: 1.2 minutes
sample_std = 1.2


# 3. Test statistic
# one-sample t test for pop. mean; pop. std unknown
t_statistic = (sample_mean - null_mean) / (sample_std / np.sqrt(sample_size))
print(f"test statistic: {round(t_statistic, 10)}")


# 4. Compare test statistic with left-tailed p-value
# significance
alpha = 0.05

# P-value
p_value = stats.t.cdf(t_statistic, df=sample_size - 1)
print(f"p value: {round(p_value, 10)}")

# 5. Interpres results
if (p_value < alpha):
  print("\nP-value < alpha")
  print("REJECT the null hypothesis:")
  print("There IS significant evidence to support the claim that the average response time for their customer support hotline is less than 5 minutes.")
else:
  print("\nP-value >= alpha")
  print("FAIL to reject the null hypothesis:")
  print("There is NOT significant evidence to support the claim that the average response time for their customer support hotline is less than 5 minutes.")
```
