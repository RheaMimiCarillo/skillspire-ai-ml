**1. If you randomly pick a smooth marble, what is the chance that it is red, i.e., $= P(\text{Red | Smooth})$?.**

Knowns:
1. $P\text{(Red) = 0.25.}$
  - 25% of *all* marbles in the bag are red.
2. $P\text{(Smooth|Red) = 0.80}$
  - 80% of *Red* marbles are smooth.
3. $P\text{(Smooth) = 0.50}$
  - 50% of *all* marbles in the bag are smooth.

$P(\text{Drawing Red | Given that the marble is smooth})$  
> $= P(\text{Smooth | Red}) \times P(\text{Red}) / P(\text{Smooth})$  
> $= (0.80 \times 0.25) / 0.50$  
> $= 0.20 / 0.50$  
> $= 0.40$
