```python
# assign csv file to a variable

worms_mustard_data = 'WormsAndMustardData_Raw.csv'
```

```python
# Read data into a dataframe
df = pd.read_csv(worms_mustard_data)

df.head()
```

# Cleaning and basic checks

```python
# Checking the shape of the dataframe
print(f"(Rows, Columns): {df.shape}\n")
df.info()
```

### Checking the shape of the data and seeing what cleaning may need to be done

```python
# Checking for NA values
print(f"Missing values: \n{df.isna().sum()}\n") # df.isna().sum()

# No missing values
print(df["Treatment"].value_counts())
```

```python
### Checking the shape of the data

print(f"Unique values: \n{df.nunique()}")

df.describe()
```

There are three treatment groups. Multiple metrics for growth are recorded including each plant's shoot mass, root mass, seed pod count + mass, flower bud count, and height.

### Descriptive Statistics

```python
# Descriptive statistics

df.describe()
```

```python
df.groupby("Treatment").mean(numeric_only=True)
```

```python
treatment_means = df.groupby("Treatment")[
    [
        "Height (Stretched)",
        "Wet Mass (g)",
        "Seed Pod Count",
        "Flower Buds",
        "Wet Mass (Shoots)",
        "Wet Mass (Roots)",
        "Wet R/S Ratio"
    ]
].mean().round(3).reset_index()

treatment_means
```

### Insights

```python
print(df.groupby("Treatment")["Height (Stretched)"].mean())

print(df.groupby("Treatment")["Seed Pod Count"].mean())

print(df.groupby("Treatment")["Flower Buds"].mean())

print(df.groupby("Treatment")["Wet Mass (g)"].mean())

```

The worm treatment group has the highest average measurements across each of the four main growth metrics in:

> Height  
> Wet Mass  
> Seed Pod Count  
> Flower Bud Count  

```python
growth_summary = df.groupby("Treatment")[
    [
        "Height (Stretched)",
        "Wet Mass (g)",
        "Seed Pod Count",
        "Flower Buds"
    ]
].mean().round(4)

growth_summary
```

```python
plt.figure(figsize=(8, 5))
sns.barplot(
    data=df,
    x="Treatment",
    y="Height (Stretched)",
    errorbar=None
)
plt.title("Average Mustard Plant Height by Treatment")
plt.xlabel("Treatment")
plt.ylabel("Average Stretched Height")
plt.show()
```

The tunnel treatment group has lowest total average mass, yet has the highest root to shoot ratio (0.0224).

```python
biomass_summary = df.groupby("Treatment")[
    [
        "Wet Mass (g)",
        "Wet Mass (Shoots)",
        "Wet Mass (Roots)",
        "Wet R/S Ratio"
    ]
].mean().round(4)

biomass_summary
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(
    data=df,
    x="Treatment",
    y="Wet Mass (g)"
)
plt.title("Total Wet Mass by Treatment")
plt.xlabel("Treatment")
plt.ylabel("Wet Mass (g)")
plt.show()
```

```python
biomass = df.groupby("Treatment")[
    ["Wet Mass (Shoots)", "Wet Mass (Roots)"]
].mean().reset_index()

biomass_long = biomass.melt(
    id_vars="Treatment",
    var_name="Plant Part",
    value_name="Average Wet Mass (g)"
)

plt.figure(figsize=(9, 5))
sns.barplot(
    data=biomass_long,
    x="Treatment",
    y="Average Wet Mass (g)",
    hue="Plant Part"
)
plt.title("Average Shoot and Root Wet Mass by Treatment")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(df["Height (Stretched)"], bins=10, kde=True)
plt.title("Distribution of Mustard Plant Height")
plt.xlabel("Stretched Height")
plt.ylabel("Number of Plants")
plt.show()
```
