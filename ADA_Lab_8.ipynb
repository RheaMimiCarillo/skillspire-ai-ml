#Part 2: Correlation Coefficients

Interpret the following in terms of mean, std, and correlation coefficient:

```python
import numpy as np

# Manually create a dataset with 15 observations
coffee_consumption = [2, 4, 1, 5, 3, 1, 3, 2, 4, 5, 1, 4, 3, 2, 5]
productivity_scores = [60, 75, 50, 80, 70, 55, 65, 60, 72, 85, 48, 78, 68, 58, 90]
```

```python
# Coffee vs Productivity

import pandas as pd

print(f"Coffee Mean: {np.mean(coffee_consumption)}")
print(f"Productivity Mean: {np.mean(productivity_scores)}")

print(f"\nCoffee Median: {np.median(coffee_consumption)}")
print(f"Productivity Median: {np.median(productivity_scores)}")

print(f"\nCoffee Standard Deviation: {np.std(coffee_consumption).round(2)}")
print(f"Productivity Standard Deviation: {np.std(productivity_scores).round(2)}")


print(f"\nCorrelation Coefficient:\n{np.corrcoef(coffee_consumption, productivity_scores).round(2)}")

coffee_productivity = {
    "Coffee Consumption": coffee_consumption,
    "Productivity Score": productivity_scores
}


df = pd.DataFrame(coffee_productivity)

df
```

### Coffee vs. Productivity Interpretation

Mean Coffees consumed is **3.0** with a standard deviation of **1.41**. This means that the average coffee's consumed is 3.0 coffees, give or take 1.41 coffees.

Mean Productivity score is **67.6** with a standard deviation of **12.15**. This means that participant score centered around 67.6, give or take 12.15 points.

The correlation coefficient of +0.97 indicates a **strong correlation** between Coffee Consumption & Productivity Scores

#Part 3: Probability

**1. Find the probability of getting a numbered card when a card is drawn from the pack of 52 cards.**

> $P(\text{Drawing a numbered card from a standard 52-card deck})$  
> $= P(\text{Drawing 2, 3, 4, 5, 6, 7, 8, 9, 10})$  
> $= P(\frac{36}{52})$  
> $= 0.69230769$

There are 36 numbered cards in a standard 52-card deck, so the probability for pulling a numbered card from a freshly-shuffled deck is $\frac{36}{52}$, or **0.69230769**.

**2. You have a bag containing 8 red balls, 5 green balls, and 3 blue balls. If you randomly draw two balls without replacement, what is the probability that both balls are green?**

> $P(\text{Picking two green balls in a row without replacement})$  
> $P(\frac{5}{8 + 5 + 3}) \times P(\frac{4}{8 + 4 + 3})$  
> $= P(\frac{5}{16}) \times P(\frac{4}{15})$  
> $= 0.3125 \times 0.26666667$  
> $= 0.08333333$

**3. You roll a six-sided die. What is the probability of rolling an even number or a number greater than 4?**

> $P(\text{Even}) \cup P(\text{5 or 6}) = P(\{2, 4, 5, 6\})$  
> $= \frac{4}{6} = \frac{2}{3}$  
> $\approx 0.667$

6 is both even and greater than 4, so it's only counted once for this set.
